```python
import svg
from dataclasses import dataclass
from IPython.display import SVG, display
```

```python
@dataclass
class Generator:
    size: int

    def generate(self) -> svg.SVG:
        return svg.SVG(
            width=self.size,
            height=self.size,
            xmlns='http://www.w3.org/2000/svg',
            #elements=list(self.iter_elements()),
        )
```

```python
generator = Generator(
    size=100
)
```

```python
empty = generator.generate()
```

```python
def line_with_circles(start, end, circle_radius=20, count=10, stroke="black"):
    """Create a horizontal or vertical line with evenly spaced circles."""
    x1, y1 = start
    x2, y2 = end

    if x1 != x2 and y1 != y2:
        raise ValueError("The line must be horizontal or vertical.")
    if start == end:
        raise ValueError("Start and end points must differ.")
    if count < 1:
        raise ValueError("count must be at least 1.")
    if circle_radius <= 0:
        raise ValueError("circle_radius must be positive.")

    elements = [
        #svg.Line(x1=x1, y1=y1, x2=x2, y2=y2, stroke=stroke)
    ]

    for i in range(count):
        t = i / (count - 1) if count > 1 else 0.5
        x = x1 + t * (x2 - x1)
        y = y1 + t * (y2 - y1)

        elements.append(
            svg.Circle(
                cx=x,
                cy=y,
                r=circle_radius,
                fill="none",
                stroke=stroke,
            )
        )

    result = svg.SVG(
        width=generator.size,
        height=generator.size,
        xmlns="http://www.w3.org/2000/svg",
        elements=elements,
    )

    #display(SVG(str(result)))
    return result


line_with_circles(
    start=(50, 150),
    end=(250, 150),
    circle_radius=20,
    count=10,
)
```

```python
def line_with_circles(
    start,
    end,
    circle_radius=20,
    count=10,
    stroke="black",
    elements=None,
):
    """Append circles for a horizontal or vertical line to a drawing."""
    x1, y1 = start
    x2, y2 = end

    if x1 != x2 and y1 != y2:
        raise ValueError("The line must be horizontal or vertical.")
    if start == end:
        raise ValueError("Start and end points must differ.")
    if count < 1:
        raise ValueError("count must be at least 1.")
    if circle_radius <= 0:
        raise ValueError("circle_radius must be positive.")

    if elements is None:
        elements = []

    for i in range(count):
        t = i / (count - 1) if count > 1 else 0.5
        x = x1 + t * (x2 - x1)
        y = y1 + t * (y2 - y1)

        elements.append(
            svg.Circle(
                cx=x,
                cy=y,
                r=circle_radius,
                fill="none",
                stroke=stroke,
            )
        )

    return svg.SVG(
        width=generator.size,
        height=generator.size,
        xmlns="http://www.w3.org/2000/svg",
        elements=elements,
    )


elements = []
```

```python

drawing = line_with_circles(
    start=(10, 10),
    end=(10, 90),
    circle_radius=10,
    count=10,
    elements=elements,
)

drawing = line_with_circles(
    start=(25, 50),
    end=(60, 50),
    circle_radius=10,
    count=5,
    elements=elements,
)

drawing = line_with_circles(
    start=(25, 10),
    end=(75, 10),
    circle_radius=10,
    count=7,
    elements=elements,
)

display(SVG(str(drawing)))
```
